```python
import pandas as pd
from PIL import Image
import os
from glob import glob
import shutil
```

```python
path = '/Volumes/Seagate Portable Drive/KYLE_AI_DATABASE/Detections_inventory/Detections inventory_2023.08.29.xlsx'
data = pd.read_excel(path, engine='openpyxl', sheet_name='Wolverine')
data = data[(data['Type'] == 'COLOR') & (data['Marks'] == 'Robust marks')]
data.head()
```

```python
base_dir = '/Volumes/Seagate Portable Drive/KYLE_AI_DATABASE/A_CAMERA/Focal_species/Gulo gulo'

for index, row in data.iterrows():
    img_dir = os.path.join(base_dir, str(row['ID']), str(row['Station']), str(row['Year']), 'COLOR', 'Robust marks', str(row['Date-Time']))
    ims = sorted(glob(img_dir + '/*.JPG'))
    if len(ims) > 2:
        mid_point = len(ims)//2 - 1
        focal_im = ims[mid_point]
        new_base = focal_im.split('Gulo gulo')[1].replace(' ', '').replace('/','_')[1:]
        new_full = os.path.join('init_images', new_base)
        shutil.copy(focal_im, new_full)
    print(new_full)
        
```
