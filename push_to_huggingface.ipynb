```python
from huggingface_hub import HfApi
from datasets import load_dataset
import os
api = HfApi()
```

```python
target_repo = 'kdoherty/wolverines'
rtype = 'dataset'
```

```python
#upload readme
api.upload_file(
    path_or_fileobj="README.md",
    path_in_repo="README.md",
    repo_id=target_repo,
    repo_type=rtype,
)
```

```python
#upload test image folder
api.upload_folder(
    folder_path="data/parquet",
    path_in_repo="data",
    repo_id=target_repo,
    repo_type=rtype,
    delete_patterns=["*.parquet"],
)
```
